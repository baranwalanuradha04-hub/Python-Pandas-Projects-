```python
class Student:  # The template for a student
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def passed(self, threshold):  # Method to check if student passed
        return self.mark >= threshold

threshold_mark = 75
students = []

for index, row in df.iterrows():  # Go through the dataframe one row at a time
    s = Student(row['name'], row['mark'])
    students.append(s)

# Print every student with their pass status
pass_count = 0
fail_count = 0

for s in students:
    result = s.passed(threshold_mark)
    print(s.name, s.mark, result)
    if result:
        pass_count += 1
    else:
        fail_count += 1

print("\nTotal Passed (True):", pass_count)
print("Total Failed (False):", fail_count)
```

```python
for s in students[:5]:  # only first 5
    result = s.passed(threshold_mark)
    print(s.name, s.mark, result)
```

```python
true_rows = [s for s in students if s.passed(threshold_mark)]
```

```python
true_rows = [s for s in students if s.passed(threshold_mark)]
```

```python
false_rows = [s for s in students if not s.passed(threshold_mark)]

print("Passed:")
for s in true_rows:
   print(s.name, s.mark)

print("\nFailed:")
for s in false_rows:
   print(s.name, s.mark)
```
